## Partie 2 - Nettoyage du texte

````python
text ="Hey! These days, studying NLP has become absolutely fascinating, don’t you think? It’s amazing to see how something as natural as human language can be transformed into data that machines can understand and work with."
print("Texte original :")
print(text)
````

````python
# Mise en minuscules
text_lower = text.lower()
print("Texte en minuscules :")
print(text_lower)
````

````python
text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))
print("Texte nettoyé:")
print(text_clean)
````

### Questions

Pourquoi est-il important de mettre le texte en minuscules ?

Parce qu'un même mot peut apparaître avec des casses différentes selon sa position dans la phrase si il est en début de phrase, nom propre, majuscules d'emphase.Mettre le texte en minuscules permet de regrouper toutes ces variantes sous une seule forme.


## Partie 3 - Tokenisation et Stopwords


````python
tokens = word_tokenize(text_clean, language='french')
print("Liste des tokens :")
print(tokens)
````

````python
# Liste des stopwords fournie par NLTK
stop_words = set(stopwords.words('english'))
print(f"Nombre de stopwords: {len(stop_words)}")
print(list(stop_words)[:20], "...")
````

````python
tokens_filtered = [mot for mot in tokens if mot not in stop_words]
print("Tokens après suppression des stopwords :")
print(tokens_filtered)
````

````python
mot_a_ajouter = "amazing"
stop_words.add(mot_a_ajouter)

tokens_filtered = [mot for mot in tokens_filtered if mot not in stop_words]
print(f"Mot ajouté aux stopwords : '{mot_a_ajouter}'")
print("Nouvelle liste de tokens filtrés :")
print(tokens_filtered)
````

### Questions

Que représentent les stopwords ?

Les stopwords sont les mots très fréquents d'une langue qui apportent peu ou pas d'information sur le sens du texte. On les retire avant une analyse sémantique  car ils polluent les résultats sans aider à distinguer un texte d'un autre.


````python
# Calcul de la proportion de tokens supprimés
proportion_supprimee = 1 - len(tokens_filtered) / len(tokens)
print(f"Nombre de tokens avant filtrage : {len(tokens)}")
print(f"Nombre de tokens après filtrage : {len(tokens_filtered)}")
print(f"Proportion supprimée : {proportion_supprimee:.1%}")
````

## Partie 4 - Lemmatisation


````python
import spacy
````

````python
# Lemmatisation avec spacy sur les tokens filtrés

texte_a_lemmatiser = " ".join(tokens_filtered)
doc = nlp(texte_a_lemmatiser)

lemmes = [token.lemma_ for token in doc]

print("Avant lemmatisation:", tokens_filtered)
print("Après lemmatisation:", lemmes)
````

````python
# Comparaison mot à mot
for mot, lemme in zip(tokens_filtered, lemmes):
    marque = "→ changé" if mot != lemme else ""
    print(f"{mot:15s} -> {lemme:15s} {marque}")
````

### Questions

Quelle est la différence entre stemming et lemmatisation ?

- Le stemming coupe un mot en se basant sur la suppression de suffixes pour obtenir une racine.
- La lemmatisation c'Est de  ramener un mot à sa forme canonique réelle.

Pourquoi cette étape est-elle importante avant l'analyse sémantique ?

Parce qu'un même concept peut s'exprimer sous plusieurs formes grammaticales. Sans lemmatisation, ces variantes seraient comptées comme des mots différents, ce qui disperse l'information. En ramenant chaque mot à sa forme de base.
